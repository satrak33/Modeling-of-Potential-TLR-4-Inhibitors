# ADMET Prediction

This notebook uses the `admet_ai` library to predict Absorption, Distribution, Metabolism, Excretion, and Toxicity (ADMET) properties for the filtered library.

```python
import pandas as pd
from pathlib import Path
from admet_ai import ADMETModel
```

## Load Data

Load the pre-filtered SMILES from `ligands_pre_admet.smi` and remove duplicates.

```python
ligands_input_path = Path("..") / "data" / "interim" / "ligands_pre_admet.smi"
ligands_output_path = Path("..") / "data" / "processed" / "ligands_admet.csv"

df = pd.read_csv(ligands_input_path, sep=" ", engine='python', header=0, names=["COMB", "SMILES"])
df = df.drop_duplicates(subset=['SMILES'], keep='first')
df = df.reset_index(drop=True)
print(df)
```

## Run Prediction

Initialize the `ADMETModel` and predict properties for all molecules. The results are merged with the original DataFrame and saved to `ligands_admet.csv`.

```python
if input("Are you sure? [y/n] ").lower() == "y":
    smiles_data = df["SMILES"].tolist()
    
    model = ADMETModel()
    result = model.predict(smiles_data)
    final = result.reset_index(drop=True)
    final = pd.concat([df, final], axis=1)
    final.to_csv(ligands_output_path, index=False)
else:
    print("Aborted by user")
```
