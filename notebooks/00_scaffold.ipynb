```python
from pathlib import Path
import sys

PROJECT_ROOT = next((p for p in (Path.cwd(), *Path.cwd().parents) if (p / "utils" / "paths.py").is_file()), None)
if PROJECT_ROOT is None:
    raise RuntimeError("Start the notebook inside the project directory")
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from utils.paths import RAW, stage_path
from utils.runs import require_stage, stage_run
from loguru import logger
from rdkit import Chem
from rdkit.Chem import AllChem
from pathlib import Path
from loguru import logger
import sys


from utils import draw
```

```python
mol_folder = RAW
mol_path = RAW / "curcumin.sdf"
```

```python
with stage_run("00_scaffold", [mol_path], {}) as output_dir:
    mol = Chem.SDMolSupplier(mol_path)[0]
    AllChem.Compute2DCoords(mol)
    smiles = Chem.MolToSmiles(mol)

    logger.info(f"Smiles: {smiles}")

    draw(mol)
    edit_mol = Chem.RWMol(mol)

    # R4
    replacement_atom = Chem.Atom(103)
    edit_mol.ReplaceAtom(3, replacement_atom)

    # R3
    replacement_atom = Chem.Atom(102)
    edit_mol.ReplaceAtom(1, replacement_atom)
    edit_mol.RemoveAtom(26)

    # R2
    replacement_atom = Chem.Atom(101)
    edit_mol.ReplaceAtom(2, replacement_atom)
 
    # R1
    replacement_atom = Chem.Atom(100)
    edit_mol.ReplaceAtom(0, replacement_atom)
    edit_mol.RemoveAtom(25)

    rad_mol = edit_mol.GetMol()
    AllChem.Compute2DCoords(rad_mol)


    smiles = Chem.MolToSmiles(rad_mol)

    logger.info(f"RadSmiles: {smiles}")

    smiles_path = output_dir / "curcumin_scaffold.smi"

    with open(smiles_path, "w") as f:
        f.write(smiles)

    draw(rad_mol)
```
