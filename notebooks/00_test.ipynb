# Test Notebook

This notebook tests the RDKit environment and basic molecular scaffold manipulation.
It loads a curcumin molecule, visualizes it, and creates a core scaffold by replacing specific groups with dummy atoms.

```python
from rdkit import Chem, DataStructs
from rdkit.Chem import Draw, AllChem, Lipinski, Descriptors
from IPython.display import SVG, display
from pathlib import Path
from loguru import logger
import numpy as np
from tqdm.auto import tqdm
from datetime import datetime
from itertools import product
from pprint import pprint
import io
import time
import sys
import os

sys.path.append("..")
from utils import draw
```

## Load and Visualize Curcumin

Load `curcumin.sdf` from the raw data directory, compute 2D coordinates, and display the molecule.

```python
mol_folder = Path("..") / "data" / "raw"

mol = Chem.SDMolSupplier(str(mol_folder / "curcumin.sdf"))[0]
AllChem.Compute2DCoords(mol)
smiles = Chem.MolToSmiles(mol)

print(f"Smiles: {smiles}")

draw(mol)
```

## Generate Scaffold

Create a scaffold by replacing specific substituent groups with dummy atoms (R-groups) to define attachment points for generation.
- **R4 (103)**: Replaces OH at index 3
- **R3 (102)**: Replaces O at index 1, removes CH3 at index 26
- **R2 (101)**: Replaces OH at index 2
- **R1 (100)**: Replaces O at index 0, removes CH3 at index 25

```python
edit_mol = Chem.RWMol(mol)

# R4
replacement_atom = Chem.Atom(103)
edit_mol.ReplaceAtom(3, replacement_atom)

# R3
replacement_atom = Chem.Atom(102)
edit_mol.ReplaceAtom(1, replacement_atom)
edit_mol.RemoveAtom(26)

# R2
replacement_atom = Chem.Atom(101)
edit_mol.ReplaceAtom(2, replacement_atom)
 
# R1
replacement_atom = Chem.Atom(100)
edit_mol.ReplaceAtom(0, replacement_atom)
edit_mol.RemoveAtom(25)

rad_mol = edit_mol.GetMol()
AllChem.Compute2DCoords(rad_mol)

draw(rad_mol)

smiles = Chem.MolToSmiles(rad_mol)

print(smiles)

smiles_path = mol_folder / "curcumin_scaffold.smi"

with open(smiles_path, "w") as f:
    f.write(smiles)
```
